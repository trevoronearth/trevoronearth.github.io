```python
import folium

m = folium.Map(location=[39.7392, -104.9903], zoom_start=3)

import geojson

with open('west_Highland_Way.json', 'r') as file:
    geojson_data = geojson.load(file)

# Add a marker
folium.Marker(
    location=[39.7392, -104.9903],
    popup="Downtown Denver",
    tooltip="Click me!"
).add_to(m)

folium.Marker(
    location=[39.8561, -104.6737],
    popup="Denver International Airport",
    icon=folium.Icon(color="green", icon="plane")
).add_to(m)

folium.GeoJson(geojson_data).add_to(m)

m.save("map.html")

m
```
